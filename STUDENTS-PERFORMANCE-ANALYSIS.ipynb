### Identify outliers

score_columns = ["math score", "reading score", "writing score"]

Find outliers using the IQR method
for col in score_columns:
    Q1 = df[col].quantile(0.25)
    Q3 = df[col].quantile(0.75)
    IQR = Q3 - Q1

    lower_limit = Q1 - 1.5 * IQR
    upper_limit = Q3 + 1.5 * IQR

    outliers = df[(df[col] < lower_limit) | (df[col] > upper_limit)]

    print(f"{col}: {len(outliers)} potential outliers")

The IQR method was used to identify potential outliers in mathematics, reading, and writing scores. These observations were examined to understand unusual score values and variations in students’ academic performance. Since student scores naturally range from 0 to 100, potential outliers should be reviewed carefully before deciding whether to remove them.

### Remove outliers

```python
score_columns = ["math score", "reading score", "writing score"]

df_clean = df.copy()

for col in score_columns:
    Q1 = df_clean[col].quantile(0.25)
    Q3 = df_clean[col].quantile(0.75)
    IQR = Q3 - Q1

    lower_limit = Q1 - 1.5 * IQR
    upper_limit = Q3 + 1.5 * IQR

    df_clean = df_clean[
        (df_clean[col] >= lower_limit) &
        (df_clean[col] <= upper_limit)
    ]

print("Original dataset shape:", df.shape)
print("Dataset shape after removing outliers:", df_clean.shape)
```

The original dataset contained 1,000 rows and 8 columns. After adding the total score and average score features, the dataset contained 1,000 rows and 10 columns. Using the IQR method, 14 rows were removed as potential outliers, resulting in a cleaned dataset of 986 rows and 10 columns.

### 8. Correlation Between Subject Scores.
      Correlation Heatmap

```python
import seaborn as sns

correlation = df[
    ["math score", "reading score", "writing score"]
].corr()

plt.figure(figsize=(7, 5))
sns.heatmap(
    correlation,
    annot=True,
    cmap=sns.color_palette(
        ["#F5EBE0", "#B56576", "#7B2D43"],
        as_cmap=True
    ),
    vmin=-1,
    vmax=1,
    fmt=".2f"
)
plt.title("Correlation Between Subject Scores")
plt.tight_layout()
plt.show()
```

1. Perfect Self-Correlation (1.00):
When any subject is correlated with itself (e.g., Math vs. Math, Reading vs. Reading), the digital value is exactly 1.00, represented by the darkest diagonal blocks.
2. Strongest Correlation (Reading & Writing):
The correlation between reading score and writing score is the highest in the dataset at 0.95. This indicates a very strong positive relationship—students who score high in reading almost universally score high in writing.
3. Moderate-to-Strong Correlations (Math vs. Literacy):
* The correlation between math score and reading score is 0.82.
* The correlation between math score and writing score is 0.80.
* While slightly lower than the reading-writing pairing, these values still show a strong positive trend: overall academic strength tends to carry across both STEM (math) and humanities (reading/writing) subjects.
